## Project Goal:
This project seeks to analyze lung images taken at different biopsy depths and to estimate the extent of fibrosis within the lung. For the purposes of this project, fibrosis is approximated by the percentage of white pixels in black and white lung images. We then use interpolation to estimate fibrosis at depths where no direct image was measured. The long-term goal is to help inform the design of a lung biopsy device that can sample tissues at precise depths. 

## Disease Background: 

* Prevalence & incidence
   * The incidence increases with age, but approximately 3 million people are diagnosed. The prevalence for IPF in the United States is 495 out of every 100,000.
* Economic Burden
   * Due to all the resources needed for those with IPF the estimated annual cost per patient can range from 29,0000 to 37,000. On top of that, two of the available drug treatments add an additional 2,000 to 14,000 expenses. 
* Risk factors (genetic, lifestyle)
   * The main risk factors are age, lifestyle habits, sex, and family history and genetics. As you get older the risk of getting IFP increases and is mostly diagnosed in people’s 60s or 70s. Lifestyle habits such as smoking can also increase risk. In terms of sex, men are more likely to get IPF than females. Finally, if an immediate family member has had IPF you are more at risk of getting it as there are specific genes that make you more susceptible to developing IPF. Specifically, mutations in the TERT and TERC genes are common in people who have IPF.
* Societal Determinants
   * Some societal determinants can be occupation, environmental pollution, access to healthcare, and socioeconomic status. Occupation can cause an impact depending on the environment you’re working in as job such as construction can lead to dust and silica exposure. IPF is a disease that is better to catch early, and healthcare access can help with early diagnosis and better treatment. Finally, socioeconomic status can impact one’s ability to afford treatment. 
* Symptoms
   * The main symptoms are worsening shortness of breath and dry cough. Additionally, there is aching muscles and joints, clubbing of extremities, tiredness, unintended weight loss, and rapid shallow breathing. 
* Diagnosis
   * Physical exam indicators such as blue hands and feet from lack of oxygen in the blood, clubbing, and high-pitched crackles when listening to lungs. In terms of testing high resolution CT scans and lung biopsy can be used to check for lung scarring or inflation. Finally, blood tests for antibodies, lung function tests, and genetic testing are further options that can help aid in IPF diagnosis.
* Standard of care treatment(s)
   * Medicines such as nintedanib or pirfenidone can help your lungs work better. They can help prevent acute exacerbation which can help prevent deterioration and prolong survival. Additionally, antacids can help prevent stomach acid from getting into lungs from reflux which can make IPF worse. Other treatments such as oxygen therapy and ventilator support can be used to assist with breathing. Finally, if IPF is very severe a lung transplant will probably be the best option. Lung transplant is a big help with IPF, but come with possibilities for infection, organ rejection, and a lifetime of medications.
* Biological mechanisms 
   * It is a progressive lung disease marked by excessive collagen deposition and scarring in the lung tissue. This is driven by fibroblast and myofibroblast activation. This process is regulated by growth factors like TGF-B, which promotes fibrosis by stimulating ECM production. As the disease progresses, the lung becomes stiffer, imparing gas exchange and leading to symptoms like shortness of breath.

#### 1b. Define Image Depths and Caculate Pixel Percentage 

```python
depths = [
    45,
    90,
    60,
    30,
    80,
    100,
    600,
    570,
    200,
    955,
    500,
    2950,
    6600,
    780,
    55,
    15,
    380,
    400,
    585,
    3900,
    1500,
    7100,
    9000,
    2000,
    3350
]

# Make the lists that will be used

images = []
white_counts = []
black_counts = []
white_percents = []

# Build the list of all the images you are analyzing

for filename in filenames:
    img = cv2.imread(filename, 0)
    images.append(img)

# For each image (until the end of the list of images), calculate the number of black and white pixels and make a list that contains this information for each filename.

for x in range(len(filenames)):
    _, binary = cv2.threshold(images[x], 127, 255, cv2.THRESH_BINARY)

    white = np.sum(binary == 255)
    black = np.sum(binary == 0)

    white_counts.append(white)
    black_counts.append(black)

# Print the number of white and black pixels in each image.

print(colored("Counts of pixel by color in each image", "yellow"))
for x in range(len(filenames)):
    print(colored(f"White pixels in image {x}: {white_counts[x]}", "white"))
    print(colored(f"Black pixels in image {x}: {black_counts[x]}", "black"))
    print()

# Calculate the percentage of pixels in each image that are white and make a list that contains these percentages for each filename

for x in range(len(filenames)):
    white_percent = (
        100 * (white_counts[x] / (black_counts[x] + white_counts[x])))
    white_percents.append(white_percent)

# Print the filename (on one line in red font), and below that line print the percent white pixels and depth into the lung where the image was obtained

print(colored("Percent white px:", "yellow"))
for x in range(len(filenames)):
    print(colored(f'{filenames[x]}:', "red"))
    print(f'{white_percents[x]}% White | Depth: {depths[x]} microns')
    print()

```

#### 1c. Write and Print Data to CSV File

```python

'''Write your data to a .csv file'''

# Create a DataFrame that includes the filenames, depths, and percentage of white pixels
df = pd.DataFrame({
    'Filenames': filenames,
    'Depths': depths,
    'White percents': white_percents
})
# Write that DataFrame to a .csv file
df.to_csv('Percent_White_Pixels.csv', index=False)

print("The .csv file 'Percent_White_Pixels.csv' has been created.")

```

#### 1d. Interpolate and Plot Graphs 

```python
interpolate_depth = float(input(colored(
    "Enter the depth at which you want to interpolate a point (in microns): ", "yellow")))

# Create the interpolation function by organizing the different data architectures to the corresponding x and y axis.
x = depths 
y = white_percents

# You can also use 'quadratic', 'cubic', etc. We chose to use quadratic for our interpolation
i = interp1d(x, y, kind='quadratic')
interpolate_point = i(interpolate_depth)
# Print a statement to ensure that the interpolation had occurred.
print(colored(
    f'The interpolated point is at the x-coordinate {interpolate_depth} and y-coordinate {interpolate_point}.', "green"))

# Create two lists that append the interpolated point and its white pixel percentage.
depths_i = depths[:]
depths_i.append(interpolate_depth)
white_percents_i = white_percents[:]
white_percents_i.append(interpolate_point)


# make two plots: one that doesn't contain the interpolated point, just the data calculated from your images, and one that also contains the interpolated point (shown in red)
fig, axs = plt.subplots(2, 1)

# Make the first plot. Plot all points calculated by the code with its white pixel percentage. This doesn't consist of the interpolated point.
axs[0].scatter(depths, white_percents, marker='o', linestyle='-', color='blue') # Label all points with blue
axs[0].set_title('Plot of depth of image vs percentage white pixels')
axs[0].set_xlabel('depth of image (in microns)')
axs[0].set_ylabel('white pixels as a percentage of total pixels')
axs[0].grid(True)


# Make the second plot. Plot all points calculated by the code with its white pixel percentage. This one also has the interpolated point (shown in red).
axs[1].scatter(depths_i, white_percents_i, marker='o',
               linestyle='-', color='blue') # Label all points with blue
axs[1].set_title(
    'Plot of depth of image vs percentage white pixels with interpolated point (in red)')
axs[1].set_xlabel('depth of image (in microns)')
axs[1].set_ylabel('white pixels as a percentage of total pixels')
axs[1].grid(True)
# Highlight the interpolated point
axs[1].scatter(depths_i[len(depths_i)-1], white_percents_i[len(white_percents_i)-1],
               color='red', s=100, label='Highlighted point')


# Adjust layout to prevent overlap
plt.tight_layout()
# Show both graphs simultaneously
plt.show()
```

```python
# Here's a function that looks at the average white pixel percentage over the 25 images we analyzed for.
# Based on the average percentage, it provides a suggested plan of treatment that the individual can take to be safe.

def treatmentOptions():
    averageNumber = 0
    denominator = len(white_percents)
    for percentages in white_percents:
        averageNumber += percentages

    # calculate average white pixel percentage.
    averageNumber/=denominator

    # Provide a key response to the user to guide them with navigating this disease.
    if averageNumber < 0.8 and averageNumber > 0.5:
        print(colored("Treatment: Antibiotics with minimally aggressive protocol.", "yellow"))
    elif averageNumber > 0.85:
        print(colored("Treatment: Aggressive antibiotics. Condition is critical.", "red"))
    elif averageNumber < 0.5 and averageNumber > 0.3:
        print(colored("Treatment: Condition is okay. Consider oxygen therapy for proper breathing.", "lightgreen"))
    elif averageNumber < 0.3:
        print(colored("Treatment: Condition is stable for now! Consider doctor checkups if breathing gets worse.", "green"))

treatmentOptions()
```
